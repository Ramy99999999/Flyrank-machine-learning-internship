## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a ranking problem because the goal is to prioritize content pages by their potential refresh opportunity. The output helps the content team focus on the pages that should be investigated first instead of reviewing thousands of pages.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd

repo_path = "/content/Flyrank-machine-learning-internship"

if not os.path.exists(repo_path):
    !git clone -q https://github.com/Ramy99999999/Flyrank-machine-learning-internship.git

df = pd.read_csv(
    f"{repo_path}/data/raw/content_refresh_anonymized.csv"
)

print(f"Loaded {len(df):,} rows and {len(df.columns)} columns.")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

I would predict whether a content page is showing signs of declining performance. This is a proxy for refresh opportunity because a declining page may deserve investigation and a potential refresh, but decline does not guarantee that a refresh will improve performance. The proxy label is defined from the observed `trend_direction` field: pages with `trend_direction = "down"` are labeled as declining.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

df[["content_id", "trend_direction", "is_declining_label"]].head(10)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I would use Precision@K as the success metric. It measures how many of the top K pages ranked by the model are actually useful opportunities. This fits the content team's workflow because they can only investigate a limited number of pages, so having useful opportunities near the top of the ranking is more important than overall accuracy.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df["is_declining_label"].value_counts()
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row represents one content page, identified by content_id. Each row contains measurable characteristics of that page, such as search volume, CTR, average position, content age, and other performance/content features.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would require manually chosen thresholds for things like CTR, position, age, and impressions. However, refresh opportunity depends on multiple signals together, and the useful pattern may not be captured by one simple threshold. ML can learn these relationships from data and produce a more flexible prioritization.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
df[[
    "search_volume",
    "ctr",
    "avg_position",
    "content_age_days",
    "days_since_last_update"
]].describe()
```
